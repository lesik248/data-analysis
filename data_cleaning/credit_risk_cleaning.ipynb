```python
df.isna().sum()
```

## Binary features = ['Foreign_Worker', 'Class']

```python
df['Foreign_Worker'].value_counts()
```

```python
df['Foreign_Worker'] = df['Foreign_Worker'].map({'yes': 1, 'no': 0})
df['Foreign_Worker'].dtype
```

```python
df['Class'].value_counts()
```

```python
df['Class'] = df['Class'].map({'good': 1, 'bad': 0})
df['Class'].dtype
```

## Changing dtypes

```python
likely_int = ['Duration', 'Residence_Since', 'Age', 'Existing_Credits', 'Num_Dependents']
```

```python
for col in likely_int:
    df[col] = df[col].astype(int)
    print(f'{col} dtype is {df[col].dtype}')
```

## One-hot encoding of categorical columns

```python
likely_cat = ['Checking_Status', 'Personal_Status', 'Other_Parties', 'Property_Magnitude', 'Housing', 'Job']
```

### Check cardinality

```python
for col in likely_cat:
    print(df[[col]].nunique())
    print('------------------------')
```

```python
df = pd.get_dummies(df, columns=likely_cat)
df.info()
```

```python
df.isna().sum().sum()
```

```python
df.to_csv('../output/credit_customers_cleaned.csv', index=False)
```
